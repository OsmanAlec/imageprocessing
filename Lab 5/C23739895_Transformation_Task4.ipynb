```python
import cv2
from matplotlib import pyplot as plt
import numpy as np

O = cv2.imread("orange.png")

kernel = np.array([
    [0, -1, 0],
    [-1,  5, -1],
    [0, -1, 0]
], dtype=float)

F = cv2.filter2D(O, ddepth=-1, kernel=kernel)
Diff = cv2.subtract(O, F)

cv2.imshow("image", F)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)

cv2.imshow("image", Diff)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)


```
