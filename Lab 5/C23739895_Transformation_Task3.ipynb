```python
import cv2
from matplotlib import pyplot as plt
import numpy as np

O = cv2.imread("orange.png")
W = cv2.imread("water.jpg")

M = cv2.multiply(O, W, scale=0.1)
A = cv2.add(O, W)
AW = cv2.addWeighted(O, 0.5, W, 0.5, 0.0)

cv2.imshow("image", M)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)

cv2.imshow("image", A)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
cv2.imshow("image", AW)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)


```
