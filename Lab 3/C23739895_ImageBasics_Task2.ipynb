```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
import easygui
```

```python
I = cv2.imread("Road.jpg")
Original = I.copy()
```

```python
def draw(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        left_x = x-101 if x-101 >= 0 else 0
        right_x = x+100 if x+100 <= I.shape[1] else I.shape[1]
        top_y = y-101 if y-101 >= 0 else 0
        bottom_y = y+100 if y+100 <= I.shape[0] else I.shape[0]
        YUV = cv2.cvtColor(I, cv2.COLOR_BGR2YUV)
        I[top_y:bottom_y,left_x:right_x,:] = YUV[top_y:bottom_y,left_x:right_x,:]
        cv2.rectangle(img = I, pt1 = (left_x,top_y), pt2 = (right_x,bottom_y),
                    color = (0, 0, 255),
                   thickness = 5)

        cv2.imshow("image", I)

cv2.namedWindow("image") 
cv2.setMouseCallback("image", draw) 
cv2.imshow("image", I)
key = cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
            
```
