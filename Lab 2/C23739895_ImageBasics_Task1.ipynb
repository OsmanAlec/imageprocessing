```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

I = cv2.imread('Road.jpg')
```

```python
# Showing BGR in RGB matplotlib
plt.imshow(I)
```

```python
# Convert to normal rgb so it's compatible with matplotlib
rgb = cv2.cvtColor(I, cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
```

```python
G = cv2.cvtColor(I, cv2.COLOR_BGR2GRAY)
plt.imshow(G, cmap='gray')
```

```python
YUV = cv2.cvtColor(rgb, cv2.COLOR_BGR2YUV)
Y = YUV[:,:,0] # The first channel (channel 0) is Y
```

```python
cv2.imshow("luminescence", Y)
key = cv2.waitKey(0)
```

```python
cv2.imshow('grayscale', G)
key = cv2.waitKey(0)
```

```python


plt.subplot(1, 2, 1)
plt.imshow(G, cmap='gray')
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(Y, cmap='gray')
plt.axis("off")

plt.show()
```
