```python
import cv2
from matplotlib import pyplot as plt
import numpy as np

I = cv2.imread("Monalisa.png")

(h, w) = I.shape[:2]
quarter = I[h//2:h, w//2:w]

(h,w) = quarter.shape[:2]
cx = w//2
cy = h//2
d = 45
s = 1

M = cv2.getRotationMatrix2D(center=(cx,cy), angle=d, scale=0.5)
R = cv2.warpAffine(quarter, M=M, dsize=(w,h))

cv2.imshow("image", R)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)

```
