```python
import cv2
from matplotlib import pyplot as plt

I = cv2.imread("wartime.jpg")
```

```python
Values = I.ravel()

plt.subplot(1, 2, 1)
plt.imshow(I)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.hist(Values,bins=256,range=[0,256]);

plt.show()
```

```python
G = cv2.imread("wartime.jpg", 0)
H = cv2.equalizeHist(G)
Values2 = H.ravel()


plt.subplot(1, 2, 1)
plt.imshow(H, cmap="grey")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.hist(Values2,bins=256,range=[0,256]);

plt.show()
```

```python
M = cv2.imread("Monalisa.png")


img_yuv = cv2.cvtColor(M, cv2.COLOR_BGR2YUV)

img_yuv[:,:,0] = cv2.equalizeHist(img_yuv[:,:,0])

img_output = cv2.cvtColor(img_yuv, cv2.COLOR_YUV2BGR)

plt.subplot(1, 2, 1)
plt.imshow(M, cmap="grey")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_output, cmap="grey")
plt.axis("off")

plt.show()
```
